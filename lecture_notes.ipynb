# Making a basic figure with matplotlib

```python
#scatter plot

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y)
```

```python
#bar plot 

fig, ax = plt.subplots(figsize = (5,3))
ax.bar(x,y)
```

```python
#changes orientation of bar plot

fig, ax = plt.subplots(figsize = (5,3))
ax.barh(x,y)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y) # generates a line graph
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.hist(y) #just counts frequency of y values
```

```python
#add labels to figure

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
#changing font and font size, and saving within dictonaries

font1 = {'family':'sans-serif','color':'blue','size':20}

font2 = {'family':'monospace','color':'green','size':14}
```

```python
#implementing font, font size changes

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

ax.set_title('Total growth over time ', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
# April 29, 2025

#Legends

#sample data
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

```

```python
#line plot with both y variables on the same axes
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1,y2) #alt solution: two separate lines with (x,y1) and (x,y2)
```

```python
# add legend to line plot

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc = 'lower right')
```

```python
#add arguments to the legend object

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right',
          frameon = True,
          fontsize = 12,
          ncol = 2,
          shadow = True)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x, y1, label = 'Person 1')
ax.plot(x, y2, label = 'Person 2')
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1))
```

```python
#Annotations with the text() function

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.text(10,95, 'This value is important!',
        ha = 'center',  #alignment of text
        color = 'red',  #modify font color
        size = 20)      #modify font size
```

```python
#positioning text annotations

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)", transform = ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform = ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform = fig.transFigure)
```

```python
#Annotating with arrows

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.annotate('This is important!', xy = (10,95), xytext = (20,94),
            arrowprops = dict(facecolor = 'black', shrink = 0.05))
```

```python
#modifying arrows

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.annotate('This is important!', xy = (10,95), xytext = (20,94),
            arrowprops = dict(arrowstyle = 'wedge', color = 'hotpink'))
```

# Axes Labels

```python
#Removing tick marks and labels

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# Limiting tick marks

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = 'Person 1')
ax.scatter(x, y2, label = 'Person 2')
ax.legend(loc = 'lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #set tick marks at every multiple of 5
```

```python
#May 1, 2025: subplots and combining visualizations

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
#introducing subplots
#plot two plots next to each other (need to define multiple axes and their relative positions)
plt.style.use('default')
font1 = {'family':'sans-serif', 'color': 'red', 'size': 12}
 
fig, (ax1, ax2) = plt.subplots(ncols = 2, nrows = 1, figsize = (7, 3))

ax1.scatter(x1, y1)
ax2.bar(x2, y2)

ax2.set_ylabel('Amount', fontdict= font1)
ax2.set_xlabel('Names', fontdict= font1)
fig.tight_layout()
```

```python
#subplots without a grid arrangement

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize = (7,4),
                                   layout = 'constrained') #add this to make labels fit

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2, y2)
someaxes['ax3'].plot(x1, y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# multiple visualizations on one axes

x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize = (7,3))

ax.bar(x, y1, color = 'indigo')
ax.plot(x, y2, color ='red')
```

```python
#Add error information

y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x, y2, yerr = y2_sd, fmt = "none", ecolor = 'indigo',
            elinewidth = 4, capsize = 6, capthick = 4, errorevery = 2)
```

```python
from PIL import Image
from io import BytesIO
```

```python
#using images from the internet

response = requests.get('http://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
#add image to a plot
fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x, y2, color = 'red')

#overlay a new axes on our figure to act as a contaner for our image
ax_image = fig.add_axes([0.1, # x coordinate on figure
                         0.11, # y coordinate on figure
                         0.15, # image width
                         0.35]) #image height

ax_image.imshow(image)
ax_image.axis('off')
```
